````python
from google.colab import files
uploaded = files.upload()
import os

pdf_path = list(uploaded.keys())[0]
print("Reading file:", pdf_path)
import pdfplumber

all_text = ""

with pdfplumber.open(pdf_path) as pdf:
    for page in pdf.pages:
        page_text = page.extract_text()
        if page_text:
            all_text += page_text + "\n"

print(all_text[:1000])
import re

def extract_details(text):
    details = {}

    # VIN (17 characters)
    vin = re.search(r'\b[A-HJ-NPR-Z0-9]{17}\b', text)
    details["VIN"] = vin.group() if vin else "Not found"

    # Car Model & Make (simple heuristic)
    model = re.search(r'(Toyota|Honda|Hyundai|Maruti|Tata|Ford|BMW|Audi)\s+\w+', text, re.IGNORECASE)
    details["Car_Model"] = model.group() if model else "Not found"

    # Year
    year = re.search(r'\b(20[0-2][0-9])\b', text)
    details["Year"] = year.group() if year else "Not found"

    # Interest Rate
    interest = re.search(r'(\d+(\.\d+)?\s*%)', text)
    details["Interest_Rate"] = interest.group() if interest else "Not found"

    # Monthly Payment
    payment = re.search(r'₹\s?\d{3,6}', text)
    details["Monthly_Payment"] = payment.group() if payment else "Not found"

    return details


extracted_details = extract_details(all_text)
extracted_details

````

````python
%%writefile backend.py

import json
import os
from datetime import datetime

import requests
import google.generativeai as genai

# ================= CONFIG =================

# Retrieve Gemini API key from environment variable (for Streamlit deployment)
GOOGLE_API_KEY = os.getenv('GOOGLE_API_KEY')

if not GOOGLE_API_KEY:
    raise RuntimeError(
        "GOOGLE_API_KEY environment variable not set. "
        "Please set it before running the app."
    )

genai.configure(api_key=GOOGLE_API_KEY)
model = genai.GenerativeModel("gemini-1.5-flash")

# Sample VIN (not used by Streamlit directly, UI passes VIN)
VIN = "1HGCM82633A004352"


# ================= SAFE JSON PARSER =================

def extract_json(text: str):
    text = text.strip()
    if text.startswith("```"):
        parts = text.split("```")
        if len(parts) >= 2:
            text = parts[1].strip()
    start = text.find("{")
    end = text.rfind("}") + 1
    if start == -1 or end == 0:
        raise ValueError("No JSON found in model response")
    return json.loads(text[start:end])


# ================= LLM FINANCIAL DATA =================

def generate_synthetic_financials(make, model_name, year, fuel_type, vehicle_age):
    prompt = (
        f"Generate realistic leasing & finance data for:\n"
        f"Vehicle: {year} {make} {model_name}\n"
        f"Fuel: {fuel_type}\n"
        f"Age: {vehicle_age} years\n\n"
        "Return ONLY JSON:\n"
        "{\n"
        "  \"interest_rate_percent\": number,\n"
        "  \"lease_term_months\": integer,\n"
        "  \"monthly_payment_usd\": number,\n"
        "  \"down_payment_usd\": number,\n"
        "  \"residual_value_percent\": number,\n"
        "  \"mileage_allowance_miles\": integer,\n"
        "  \"early_termination_clause\": \"text\",\n"
        "  \"purchase_option_usd\": number,\n"
        "  \"maintenance_responsibilities\": \"text\",\n"
        "  \"warranty_insurance_coverage\": \"text\",\n"
        "  \"penalties_late_fee_clauses\": \"text\"\n"
        "}"
    )

    response = model.generate_content(prompt)
    return extract_json(response.text)


# ================= VIN DECODE =================

def decode_vin(vin):
    url = f"https://vpic.nhtsa.dot.gov/api/vehicles/DecodeVin/{vin}?format=json"
    r = requests.get(url, timeout=15)
    r.raise_for_status()
    vehicle = {}
    for item in r.json().get("Results", []):
        if item["Value"] and item["Value"] != "Not Applicable":
            vehicle[item["Variable"]] = item["Value"]
    return vehicle


# ================= COMPLAINTS =================

def get_complaints(make, model_name, year):
    model_clean = model_name.replace(" ", "").replace("-", "")
    url = (
        "https://api.nhtsa.gov/complaints/complaintsByVehicle"
        f"?make={make}&model={model_clean}&modelYear={year}"
    )
    try:
        r = requests.get(url, timeout=15)
        return r.json().get("results", []) if r.status_code == 200 else []
    except Exception:
        return []


def format_complaints(complaints):
    return [
        {
            "date_received": c.get("dateReceived"),
            "component": c.get("component"),
            "summary": c.get("summary"),
            "crash": c.get("crash"),
            "fire": c.get("fire"),
        }
        for c in complaints
    ]


# ================= RECALLS =================

def get_recalls(vin):
    try:
        r = requests.get(
            f"https://api.nhtsa.gov/recalls/recallsByVehicle?vin={vin}",
            timeout=15,
        )
        return r.json().get("results", []) if r.status_code == 200 else []
    except Exception:
        return []


# ================= SAFETY =================

def get_accident_details(make, model_name, year):
    try:
        r = requests.get(
            f"https://api.nhtsa.gov/SafetyRatings/modelyear/{year}/make/{make}/model/{model_name}",
            timeout=15,
        )
        vehicles = r.json().get("Results", [])
        if not vehicles:
            return []
        vid = vehicles[0]["VehicleId"]
        return requests.get(
            f"https://api.nhtsa.gov/SafetyRatings/VehicleId/{vid}",
            timeout=15,
        ).json().get("Results", [])
    except Exception:
        return []


# ================= ANALYTICS =================

def compute_fairness_score(fin):
    score = 100
    if fin["interest_rate_percent"] > 14:
        score -= 20
    if fin["monthly_payment_usd"] > 750:
        score -= 15
    if fin["down_payment_usd"] > 4000:
        score -= 10
    if fin["residual_value_percent"] < 45:
        score -= 15
    return max(0, score)


def detect_red_flags(fin):
    flags = []
    if fin["interest_rate_percent"] > 15:
        flags.append("High interest rate")
    if fin["mileage_allowance_miles"] <= 10000:
        flags.append("Low mileage allowance")
    return flags


# ================= NEGOTIATION ASSISTANT (ADVANCED) =================

def negotiation_assistant(fin, fairness, red_flags, complaints):
    reasoning = []
    savings = {}
    leverage = "Low"

    if fin["interest_rate_percent"] > 14:
        reasoning.append("Interest rate is above market average")

    if fin["monthly_payment_usd"] > 700:
        reasoning.append("Monthly payment is higher than expected")

    if fin["mileage_allowance_miles"] <= 10000:
        reasoning.append("Low mileage increases overage risk")

    if len(complaints) >= 3:
        leverage = "High"
        reasoning.append("Multiple customer complaints give negotiation leverage")

    reduced_payment = round(fin["monthly_payment_usd"] * 0.9, 2)
    savings["monthly_savings_usd"] = round(
        fin["monthly_payment_usd"] - reduced_payment, 2
    )
    savings["total_savings_usd"] = round(
        savings["monthly_savings_usd"] * fin["lease_term_months"], 2
    )

    decision = (
        "WALK AWAY"
        if fairness < 50
        else "NEGOTIATE"
        if fairness < 75
        else "ACCEPT"
    )

    confidence = round(min(0.95, max(0.4, fairness / 100)), 2)

    script = (
        "Based on market comparisons and reported vehicle issues, I would like to "
        "renegotiate the lease terms, specifically requesting a lower interest rate, "
        "reduced monthly payment, and improved mileage allowance."
    )

    return {
        "deal_quality": "Good"
        if fairness >= 80
        else "Moderate"
        if fairness >= 60
        else "Poor",
        "negotiation_reasoning": reasoning,
        "negotiation_leverage": leverage,
        "suggested_counter_offer": {
            "interest_rate_percent": max(fin["interest_rate_percent"] - 2, 8),
            "monthly_payment_usd": reduced_payment,
            "mileage_allowance_miles": max(
                fin["mileage_allowance_miles"], 12000
            ),
        },
        "potential_savings": savings,
        "final_decision": {
            "decision": decision,
            "confidence": confidence,
        },
        "negotiation_script": script,
    }


# ================= REPORT =================

def generate_vehicle_report(vin):
    vehicle = decode_vin(vin)

    make = vehicle.get("Make", "Unknown")
    model_name = vehicle.get("Model", "Unknown")
    year = int(vehicle.get("Model Year", datetime.now().year))
    fuel = vehicle.get("Fuel Type - Primary", "Petrol")
    age = max(0, datetime.now().year - year)

    complaints = format_complaints(get_complaints(make, model_name, year))
    recalls = get_recalls(vin)
    accidents = get_accident_details(make, model_name, year)
    finance = generate_synthetic_financials(make, model_name, year, fuel, age)

    fairness = compute_fairness_score(finance)
    red_flags = detect_red_flags(finance)

    report = {
        "vehicle_specifications": vehicle,
        "complaints_summary": {
            "total_complaints": len(complaints),
            "complaint_details": complaints[:10],
        },
        "recalls_count": len(recalls),
        "accident_details": accidents,
        "synthetic_financials": finance,
        "contract_fairness_score": fairness,
        "red_flags": red_flags,
        "negotiation_assistant": negotiation_assistant(
            finance, fairness, red_flags, complaints
        ),
        "generated_at": datetime.now().isoformat(),
    }

    with open("vehicle_full_report.json", "w", encoding="utf-8") as f:
        json.dump(report, f, indent=2)

    print("✅ vehicle_full_report.json generated successfully")
    return report  # also return it so Streamlit can use it directly


if __name__ == "__main__":
    generate_vehicle_report(VIN)
````
